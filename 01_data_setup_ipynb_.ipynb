```python
import os

def count_images(base_folder, class_map):
    counts = {}
    for label, folder in class_map.items():
        full_path = os.path.join(base_folder, folder)
        if os.path.exists(full_path):
            counts[label] = len(os.listdir(full_path))
        else:
            counts[label] = "FOLDER NOT FOUND"
    return counts
```

```python
import os, shutil, random

random.seed(42)  # so results are reproducible across your team

ready_path = '/content/drive/MyDrive/plant_disease_project/data_ready'
for split in ['train', 'val', 'field_test']:
    for cls in pv_classes.keys():
        os.makedirs(f'{ready_path}/{split}/{cls}', exist_ok=True)

print("Structure created.")
```

```python
for label, folder in pv_classes.items():
    src = os.path.join(pv_base, folder)
    files = os.listdir(src)
    random.shuffle(files)

    train_files = files[:400]
    val_files = files[400:500]

    for f in train_files:
        shutil.copy(os.path.join(src, f), f'{ready_path}/train/{label}/{f}')
    for f in val_files:
        shutil.copy(os.path.join(src, f), f'{ready_path}/val/{label}/{f}')

    print(f"{label}: {len(train_files)} train, {len(val_files)} val")
```

```python
for label, folder in pd_classes.items():
    train_src = os.path.join(pd_train_base, folder)
    test_src = os.path.join(pd_test_base, folder)
    dest = f'{ready_path}/field_test/{label}'

    count = 0
    for f in os.listdir(train_src):
        shutil.copy(os.path.join(train_src, f), os.path.join(dest, f))
        count += 1
    for f in os.listdir(test_src):
        shutil.copy(os.path.join(test_src, f), os.path.join(dest, f'test_{f}'))  # prefix to avoid name clashes
        count += 1

    print(f"{label}: {count} field images")
```

```python
for split in ['train', 'val', 'field_test']:
    print(f"\n{split.upper()}:")
    for cls in pv_classes.keys():
        n = len(os.listdir(f'{ready_path}/{split}/{cls}'))
        print(f"  {cls}: {n}")
```
